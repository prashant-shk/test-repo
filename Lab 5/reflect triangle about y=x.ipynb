```python
import numpy as np
import matplotlib.pyplot as plt

t = np.array([
    [0, 2, 1],   
    [0, 0, 2],  
    [1, 1, 1]   
])

m = np.array([
    [-1, 0, 0],
    [0, -1, 0],
    [0, 0, 1]
])

rt = m @ t

plt.figure(figsize=(5,5))

plt.plot(
    [t[0,0], t[0,1], t[0,2], t[0,0]],
    [t[1,0], t[1,1], t[1,2], t[1,0]],
    color='violet', linestyle='-', label='original triangle'
)

plt.plot(
    [rt[0,0], rt[0,1], rt[0,2], rt[0,0]],
    [rt[1,0], rt[1,1], rt[1,2], rt[1,0]],
    color='pink', label='reflected triangle'
)

plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("reflection of a triangle about origin")
plt.legend()
plt.grid()
plt.axis("equal")
plt.show()
```
