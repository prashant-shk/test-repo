```python
import numpy as np
import matplotlib.pyplot as plt

t = np.array([
    [0, 2, 1],   
    [0, 0, 2],  
    [1, 1, 1]   
])

m = 1
c = 1
th = np.arctan(m)

T1 = np.array([
    [1, 0, 0],
    [0, 1, -c],
    [0, 0, 1]
])

T2 = np.array([
    [1, 0, 0],
    [0, 1, c],
    [0, 0, 1]
])

R1 = np.array([
    [ np.cos(th),  np.sin(th), 0],
    [-np.sin(th),  np.cos(th), 0],
    [0, 0, 1]
])

R2 = np.array([
    [ np.cos(th), -np.sin(th), 0],
    [ np.sin(th),  np.cos(th), 0],
    [0, 0, 1]
])

Rx = np.array([
    [1, 0, 0],
    [0, -1, 0],
    [0, 0, 1]
])

rf = T2 @ R2 @ Rx @ R1 @ T1
rt = rf @ t

plt.figure(figsize=(6,6))

plt.plot(
    [t[0,0], t[0,1], t[0,2], t[0,0]],
    [t[1,0], t[1,1], t[1,2], t[1,0]],
    color='yellow', linestyle='-', label='original triangle'
)

plt.plot(
    [rt[0,0], rt[0,1], rt[0,2], rt[0,0]],
    [rt[1,0], rt[1,1], rt[1,2], rt[1,0]],
    color='green', linestyle='--', label='reflected triangle'
)

plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("reflection of a triangle about y = mx + c")
plt.legend()
plt.grid()
plt.axis("equal")
plt.show()
```
